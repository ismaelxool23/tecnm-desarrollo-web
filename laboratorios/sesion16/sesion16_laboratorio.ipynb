### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
import hmac
import secrets

from fastapi import Depends, FastAPI, Request
from fastapi.encoders import jsonable_encoder
from fastapi.exceptions import RequestValidationError
from fastapi.responses import JSONResponse
from fastapi.security import OAuth2PasswordBearer, OAuth2PasswordRequestForm
from starlette.exceptions import HTTPException as StarletteHTTPException

app = FastAPI(title="Tu OAuth2 Password Flow se verá así")

USUARIOS = {"ana": "secreta"}

# token -> usuario, en memoria
tokens_activos: dict[str, str] = {}

# tokenUrl indica a Swagger UI dónde pedir el token (botón "Authorize")
esquema_oauth2 = OAuth2PasswordBearer(tokenUrl="token")


def usuario_actual(token: str = Depends(esquema_oauth2)) -> str:
    """Si falta el token o el esquema no es Bearer, OAuth2PasswordBearer ya responde 401."""
    usuario = tokens_activos.get(token)
    if usuario is None:
        raise StarletteHTTPException(status_code=401, detail="Token inválido")
    return usuario


@app.post("/token")
def emitir_token(formulario: OAuth2PasswordRequestForm = Depends()):
    esperada = USUARIOS.get(formulario.username, "")
    if not esperada or not hmac.compare_digest(formulario.password, esperada):
        raise StarletteHTTPException(status_code=401, detail="Credenciales inválidas")
    token = secrets.token_urlsafe(32)
    tokens_activos[token] = formulario.username
    return {"access_token": token, "token_type": "bearer"}


@app.get("/api/perfil")
def perfil(usuario: str = Depends(usuario_actual)):
    return {"usuario": usuario}


def sobre_error(estado: int, codigo: str, mensaje: str, detalles=None) -> JSONResponse:
    cuerpo = {"error": {"codigo": codigo, "mensaje": mensaje}}
    if detalles is not None:
        cuerpo["error"]["detalles"] = detalles
    return JSONResponse(status_code=estado, content=cuerpo)


@app.exception_handler(StarletteHTTPException)
async def manejar_http(request: Request, exc: StarletteHTTPException):
    codigos = {
        401: "NO_AUTENTICADO",
        404: "NO_ENCONTRADO",
        405: "METODO_NO_PERMITIDO",
        409: "CONFLICTO",
    }
    respuesta = sobre_error(exc.status_code, codigos.get(exc.status_code, "ERROR_HTTP"), str(exc.detail))
    if exc.status_code == 401:
        respuesta.headers["WWW-Authenticate"] = "Bearer"
    return respuesta


@app.exception_handler(RequestValidationError)
async def manejar_validacion(request: Request, exc: RequestValidationError):
    detalles = jsonable_encoder(
        [
            {"campo": ".".join(str(p) for p in e["loc"][1:]), "mensaje": e["msg"]}
            for e in exc.errors()
        ]
    )
    return sobre_error(422, "VALIDACION", "Los datos enviados no son válidos", detalles)
````
