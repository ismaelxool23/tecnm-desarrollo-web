### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import Depends, FastAPI, HTTPException
from sqlalchemy import create_engine, select
from sqlalchemy.orm import DeclarativeBase, Mapped, Session, mapped_column, sessionmaker

engine = create_engine("sqlite:///biblioteca.db", connect_args={"check_same_thread": False})
SessionLocal = sessionmaker(bind=engine, expire_on_commit=False)


# ---------- Modelo ORM ----------
class Base(DeclarativeBase):
    pass


class Libro(Base):
    __tablename__ = "libros"

    id: Mapped[int] = mapped_column(primary_key=True)
    titulo: Mapped[str] = mapped_column(nullable=False)
    anio: Mapped[int] = mapped_column(nullable=False)


# ---------- Creación de tablas y datos iniciales ----------
def sembrar() -> None:
    Base.metadata.create_all(engine)
    with SessionLocal() as db:
        if db.scalar(select(Libro.id).limit(1)) is None:
            db.add_all(
                [
                    Libro(titulo="Cien años de soledad", anio=1967),
                    Libro(titulo="Rayuela", anio=1963),
                ]
            )
            db.commit()


sembrar()

app = FastAPI(title="Búsqueda segura de libros")


def get_db():
    with SessionLocal() as db:
        yield db


def libro_a_dict(libro: Libro) -> dict:
    return {"id": libro.id, "titulo": libro.titulo, "anio": libro.anio}


# ---------- Rutas ----------
# Esta ruta debe declararse ANTES que /api/libros/{libro_id};
# si no, FastAPI intentaría interpretar "buscar" como un id entero y respondería 422.
@app.get("/api/libros/buscar")
def buscar_libros(titulo: str = "", db: Session = Depends(get_db)):
    # El valor viaja como parámetro enlazado: nunca se concatena al SQL.
    consulta = (
        select(Libro)
        .where(Libro.titulo.like(f"%{titulo}%", escape="\\"))
        .order_by(Libro.id)
    )
    return {"libros": [libro_a_dict(l) for l in db.scalars(consulta).all()]}


@app.get("/api/libros")
def listar_libros(db: Session = Depends(get_db)):
    libros = db.scalars(select(Libro).order_by(Libro.id)).all()
    return {"libros": [libro_a_dict(l) for l in libros]}


@app.get("/api/libros/{libro_id}")
def obtener_libro(libro_id: int, db: Session = Depends(get_db)):
    libro = db.get(Libro, libro_id)
    if libro is None:
        raise HTTPException(status_code=404, detail="Libro no encontrado")
    return libro_a_dict(libro)
````
