### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
import asyncio

from fastapi import Depends, FastAPI, HTTPException, Response
from pydantic import BaseModel, Field
from sqlalchemy import select
from sqlalchemy.ext.asyncio import AsyncSession, async_sessionmaker, create_async_engine
from sqlalchemy.orm import DeclarativeBase, Mapped, mapped_column

DATABASE_URL = "sqlite+aiosqlite:///biblioteca.db"

engine = create_async_engine(DATABASE_URL)
SessionLocal = async_sessionmaker(engine, expire_on_commit=False)


# ---------- Modelo ORM ----------
class Base(DeclarativeBase):
    pass


class Libro(Base):
    __tablename__ = "libros"

    id: Mapped[int] = mapped_column(primary_key=True, autoincrement=True)
    titulo: Mapped[str] = mapped_column(nullable=False)
    anio: Mapped[int] = mapped_column(nullable=False)


# ---------- Inicialización de la base ----------
_listo = False
_candado = asyncio.Lock()


async def iniciar_db() -> None:
    """Crea las tablas la primera vez que se necesitan."""
    global _listo
    if _listo:
        return
    async with _candado:
        if not _listo:
            async with engine.begin() as conn:
                await conn.run_sync(Base.metadata.create_all)
            _listo = True


app = FastAPI(title="CRUD asíncrono de libros")


@app.on_event("startup")
async def al_arrancar():
    await iniciar_db()


async def get_db():
    await iniciar_db()  # respaldo por si el evento de arranque no corrió
    async with SessionLocal() as sesion:
        yield sesion


# ---------- Esquemas ----------
class LibroDatos(BaseModel):
    titulo: str = Field(min_length=1)
    anio: int


def a_dict(libro: Libro) -> dict:
    return {"id": libro.id, "titulo": libro.titulo, "anio": libro.anio}


# ---------- Rutas ----------
@app.post("/api/libros", status_code=201)
async def crear_libro(datos: LibroDatos, db: AsyncSession = Depends(get_db)):
    libro = Libro(titulo=datos.titulo, anio=datos.anio)
    db.add(libro)
    await db.commit()
    await db.refresh(libro)
    return a_dict(libro)


@app.get("/api/libros")
async def listar_libros(db: AsyncSession = Depends(get_db)):
    resultado = await db.execute(select(Libro).order_by(Libro.id))
    return {"libros": [a_dict(l) for l in resultado.scalars().all()]}


@app.get("/api/libros/{libro_id}")
async def obtener_libro(libro_id: int, db: AsyncSession = Depends(get_db)):
    libro = await db.get(Libro, libro_id)
    if libro is None:
        raise HTTPException(status_code=404, detail="Libro no encontrado")
    return a_dict(libro)


@app.put("/api/libros/{libro_id}")
async def actualizar_libro(libro_id: int, datos: LibroDatos, db: AsyncSession = Depends(get_db)):
    libro = await db.get(Libro, libro_id)
    if libro is None:
        raise HTTPException(status_code=404, detail="Libro no encontrado")
    libro.titulo = datos.titulo
    libro.anio = datos.anio
    await db.commit()
    await db.refresh(libro)
    return a_dict(libro)


@app.delete("/api/libros/{libro_id}", status_code=204)
async def eliminar_libro(libro_id: int, db: AsyncSession = Depends(get_db)):
    libro = await db.get(Libro, libro_id)
    if libro is None:
        raise HTTPException(status_code=404, detail="Libro no encontrado")
    await db.delete(libro)
    await db.commit()
    return Response(status_code=204)
````
