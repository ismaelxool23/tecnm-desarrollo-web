### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
import base64
import hashlib
import hmac
import json
import os
import time

from fastapi import Depends, FastAPI, Header, HTTPException
from pydantic import BaseModel, Field

app = FastAPI(title="API con autenticación y roles")

SECRET_KEY = os.environ.get("SECRET_KEY", "clave-solo-para-desarrollo").encode()
TOKEN_TTL = 3600  # segundos


# ---------- Utilidades de seguridad ----------
def hashear(clave: str, sal: bytes | None = None) -> str:
    sal = sal or os.urandom(16)
    h = hashlib.pbkdf2_hmac("sha256", clave.encode(), sal, 100_000)
    return sal.hex() + "$" + h.hex()


def verificar(clave: str, almacenado: str) -> bool:
    sal_hex, h_hex = almacenado.split("$")
    nuevo = hashlib.pbkdf2_hmac("sha256", clave.encode(), bytes.fromhex(sal_hex), 100_000)
    return hmac.compare_digest(nuevo.hex(), h_hex)


def _b64(datos: bytes) -> str:
    return base64.urlsafe_b64encode(datos).decode().rstrip("=")


def _unb64(texto: str) -> bytes:
    return base64.urlsafe_b64decode(texto + "=" * (-len(texto) % 4))


def crear_token(usuario: str, rol: str) -> str:
    cuerpo = _b64(json.dumps({"sub": usuario, "rol": rol, "exp": time.time() + TOKEN_TTL}).encode())
    firma = _b64(hmac.new(SECRET_KEY, cuerpo.encode(), hashlib.sha256).digest())
    return f"{cuerpo}.{firma}"


def leer_token(token: str) -> dict:
    try:
        cuerpo, firma = token.split(".")
        esperada = _b64(hmac.new(SECRET_KEY, cuerpo.encode(), hashlib.sha256).digest())
        if not hmac.compare_digest(firma, esperada):
            raise ValueError
        datos = json.loads(_unb64(cuerpo))
        if datos["exp"] < time.time():
            raise ValueError
        return datos
    except Exception:
        raise HTTPException(status_code=401, detail="Token inválido o expirado",
                            headers={"WWW-Authenticate": "Bearer"})


# ---------- "Base de datos" en memoria ----------
USUARIOS: dict[str, dict] = {
    "admin": {"usuario": "admin", "clave": hashear("admin123"), "rol": "admin"},
}


# ---------- Modelos ----------
class Credenciales(BaseModel):
    usuario: str = Field(min_length=1)
    password: str = Field(min_length=1)


# ---------- Dependencias ----------
def usuario_actual(authorization: str | None = Header(default=None)) -> dict:
    if not authorization or not authorization.lower().startswith("bearer "):
        raise HTTPException(status_code=401, detail="Falta el token",
                            headers={"WWW-Authenticate": "Bearer"})
    datos = leer_token(authorization[7:].strip())
    registro = USUARIOS.get(datos["sub"])
    if not registro:
        raise HTTPException(status_code=401, detail="Usuario inexistente")
    return registro


def solo_admin(actual: dict = Depends(usuario_actual)) -> dict:
    if actual["rol"] != "admin":
        raise HTTPException(status_code=403, detail="Se requiere rol de administrador")
    return actual


# ---------- Rutas ----------
@app.post("/api/registro", status_code=201)
def registro(datos: Credenciales):
    if datos.usuario in USUARIOS:
        raise HTTPException(status_code=409, detail="El usuario ya existe")
    # El rol nunca lo decide el cliente.
    USUARIOS[datos.usuario] = {
        "usuario": datos.usuario,
        "clave": hashear(datos.password),
        "rol": "usuario",
    }
    return {"usuario": datos.usuario, "rol": "usuario"}


@app.post("/api/token")
def token(datos: Credenciales):
    registro = USUARIOS.get(datos.usuario)
    if not registro or not verificar(datos.password, registro["clave"]):
        raise HTTPException(status_code=401, detail="Credenciales incorrectas")
    return {
        "access_token": crear_token(registro["usuario"], registro["rol"]),
        "token_type": "bearer",
    }


@app.get("/api/perfil")
def perfil(actual: dict = Depends(usuario_actual)):
    return {"usuario": actual["usuario"], "rol": actual["rol"]}


@app.get("/api/admin/panel")
def panel_admin(actual: dict = Depends(solo_admin)):
    return {"mensaje": f"Bienvenido al panel, {actual['usuario']}", "total_usuarios": len(USUARIOS)}
````
