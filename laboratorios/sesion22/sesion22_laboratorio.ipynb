## 🧩 El reto

Modela autores y libros con SQLAlchemy, una relación 1:N y endpoints que respeten la integridad referencial.

**Requisitos del contrato**

- El modelo Autor tiene id y nombre.
- El modelo Libro tiene id, titulo, anio y una llave foránea autor_id.
- La relación Autor.libros está declarada con relationship.
- POST /api/autores y POST /api/autores/{id}/libros responden 201.
- GET /api/autores/{id} devuelve el autor con sus libros anidados.

> **Pista conceptual:** En SQLAlchemy, `ForeignKey('autores.id')` y `relationship('Libro', back_populates='autor')` describen la relación. La base se crea con `Base.metadata.create_all(engine)`.

### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import Depends, FastAPI, HTTPException
from pydantic import BaseModel, Field
from sqlalchemy import Column, ForeignKey, Integer, String, create_engine
from sqlalchemy.orm import declarative_base, relationship, sessionmaker

DATABASE_URL = "sqlite:///biblioteca.db"

engine = create_engine(DATABASE_URL, connect_args={"check_same_thread": False})
SessionLocal = sessionmaker(bind=engine, autoflush=False, autocommit=False)
Base = declarative_base()


# ---------- Modelos ORM ----------
class Autor(Base):
    __tablename__ = "autores"

    id = Column(Integer, primary_key=True, index=True)
    nombre = Column(String, nullable=False)

    libros = relationship("Libro", back_populates="autor", order_by="Libro.id")


class Libro(Base):
    __tablename__ = "libros"

    id = Column(Integer, primary_key=True, index=True)
    titulo = Column(String, nullable=False)
    anio = Column(Integer, nullable=False)
    autor_id = Column(Integer, ForeignKey("autores.id"), nullable=False)

    autor = relationship("Autor", back_populates="libros")


Base.metadata.create_all(bind=engine)

app = FastAPI(title="Autores y libros")


# ---------- Esquemas ----------
class AutorNuevo(BaseModel):
    nombre: str = Field(min_length=1)


class LibroNuevo(BaseModel):
    titulo: str = Field(min_length=1)
    anio: int


# ---------- Dependencia ----------
def get_db():
    db = SessionLocal()
    try:
        yield db
    finally:
        db.close()


def libro_a_dict(libro: Libro) -> dict:
    return {
        "id": libro.id,
        "titulo": libro.titulo,
        "anio": libro.anio,
        "autor_id": libro.autor_id,
    }


# ---------- Rutas ----------
@app.post("/api/autores", status_code=201)
def crear_autor(datos: AutorNuevo, db=Depends(get_db)):
    autor = Autor(nombre=datos.nombre)
    db.add(autor)
    db.commit()
    db.refresh(autor)
    return {"id": autor.id, "nombre": autor.nombre}


@app.post("/api/autores/{autor_id}/libros", status_code=201)
def crear_libro(autor_id: int, datos: LibroNuevo, db=Depends(get_db)):
    autor = db.get(Autor, autor_id)
    if autor is None:
        raise HTTPException(status_code=404, detail="Autor no encontrado")
    libro = Libro(titulo=datos.titulo, anio=datos.anio, autor_id=autor.id)
    db.add(libro)
    db.commit()
    db.refresh(libro)
    return libro_a_dict(libro)


@app.get("/api/autores/{autor_id}")
def obtener_autor(autor_id: int, db=Depends(get_db)):
    autor = db.get(Autor, autor_id)
    if autor is None:
        raise HTTPException(status_code=404, detail="Autor no encontrado")
    return {
        "id": autor.id,
        "nombre": autor.nombre,
        "libros": [
            {"id": l.id, "titulo": l.titulo, "anio": l.anio} for l in autor.libros
        ],
    }


@app.get("/api/libros")
def listar_libros(db=Depends(get_db)):
    libros = db.query(Libro).order_by(Libro.id).all()
    return {"libros": [libro_a_dict(l) for l in libros]}
````
