### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
import hashlib
import hmac
import os
import secrets

from fastapi import Depends, FastAPI, Header, HTTPException, Request, Response

SECRET_KEY = os.environ.get("SECRET_KEY", "clave-solo-para-desarrollo").encode()
COOKIE_SECURE = os.environ.get("COOKIE_SECURE", "0") == "1"  # pon 1 en producción (HTTPS)
NOMBRE_COOKIE = "sesion"

app = FastAPI(title="API con protección CSRF")

# Usuario de ejemplo y sesiones en memoria.
USUARIOS = {"ana": "secreta"}
SESIONES: dict[str, str] = {}


# ---------- Token CSRF ----------
def _firma(sesion_id: str, nonce: str) -> str:
    return hmac.new(SECRET_KEY, f"{sesion_id}|{nonce}".encode(), hashlib.sha256).hexdigest()


def crear_token_csrf(sesion_id: str) -> str:
    nonce = secrets.token_urlsafe(16)
    return f"{nonce}.{_firma(sesion_id, nonce)}"


def token_csrf_valido(token: str | None, sesion_id: str) -> bool:
    if not token or "." not in token:
        return False
    nonce, firma = token.split(".", 1)
    return hmac.compare_digest(firma, _firma(sesion_id, nonce))


def exigir_csrf(request: Request, x_csrf_token: str | None = Header(default=None)) -> None:
    """Dependencia para métodos que modifican estado (POST, PUT, DELETE)."""
    sesion_id = request.cookies.get(NOMBRE_COOKIE, "")
    if not token_csrf_valido(x_csrf_token, sesion_id):
        raise HTTPException(status_code=403, detail="Token CSRF ausente o inválido")


# ---------- Rutas ----------
@app.post("/api/login")
def login(datos: dict, response: Response):
    usuario = datos.get("usuario")
    password = datos.get("password")
    esperada = USUARIOS.get(usuario)
    if esperada is None or not hmac.compare_digest(str(password), esperada):
        raise HTTPException(status_code=401, detail="Credenciales incorrectas")

    sesion_id = secrets.token_urlsafe(32)
    SESIONES[sesion_id] = usuario
    response.set_cookie(
        NOMBRE_COOKIE,
        sesion_id,
        httponly=True,       # JavaScript no puede leer la cookie
        samesite="lax",      # no se envía en peticiones POST entre sitios
        secure=COOKIE_SECURE,
        max_age=3600,
        path="/",
    )
    return {"usuario": usuario}


@app.get("/api/csrf")
def obtener_csrf(request: Request):
    sesion_id = request.cookies.get(NOMBRE_COOKIE, "")
    return {"csrf_token": crear_token_csrf(sesion_id)}


@app.post("/api/logout", status_code=204, dependencies=[Depends(exigir_csrf)])
def logout(request: Request):
    SESIONES.pop(request.cookies.get(NOMBRE_COOKIE, ""), None)
    respuesta = Response(status_code=204)
    respuesta.delete_cookie(NOMBRE_COOKIE, path="/")
    return respuesta
````
