### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import Depends, FastAPI, HTTPException
from sqlalchemy import ForeignKey, create_engine, select
from sqlalchemy.orm import (
    DeclarativeBase,
    Mapped,
    Session,
    mapped_column,
    relationship,
    selectinload,
    sessionmaker,
)

engine = create_engine("sqlite:///biblioteca.db", connect_args={"check_same_thread": False})
SessionLocal = sessionmaker(bind=engine, expire_on_commit=False)


# ---------- Modelos ORM ----------
class Base(DeclarativeBase):
    pass


class Autor(Base):
    __tablename__ = "autores"

    id: Mapped[int] = mapped_column(primary_key=True)
    nombre: Mapped[str] = mapped_column(nullable=False)
    libros: Mapped[list["Libro"]] = relationship(back_populates="autor", order_by="Libro.id")


class Libro(Base):
    __tablename__ = "libros"

    id: Mapped[int] = mapped_column(primary_key=True)
    titulo: Mapped[str] = mapped_column(nullable=False)
    anio: Mapped[int] = mapped_column(nullable=False)
    autor_id: Mapped[int] = mapped_column(ForeignKey("autores.id"), nullable=False)
    autor: Mapped[Autor] = relationship(back_populates="libros")


# ---------- Creación de tablas y datos iniciales ----------
def sembrar() -> None:
    Base.metadata.create_all(engine)
    with SessionLocal() as db:
        if db.scalar(select(Autor.id).limit(1)) is not None:
            return
        garcia = Autor(nombre="Gabriel García Márquez")
        cortazar = Autor(nombre="Julio Cortázar")
        garcia.libros.append(Libro(titulo="Cien años de soledad", anio=1967))
        cortazar.libros.append(Libro(titulo="Rayuela", anio=1963))
        db.add_all([garcia, cortazar])
        db.commit()


sembrar()

app = FastAPI(title="Autores y libros con carga ansiosa")


def get_db():
    with SessionLocal() as db:
        yield db


# ---------- Serialización ----------
def libro_a_dict(libro: Libro) -> dict:
    return {"id": libro.id, "titulo": libro.titulo, "anio": libro.anio, "autor_id": libro.autor_id}


def autor_a_dict(autor: Autor) -> dict:
    return {
        "id": autor.id,
        "nombre": autor.nombre,
        "libros": [{"id": l.id, "titulo": l.titulo, "anio": l.anio} for l in autor.libros],
    }


# ---------- Rutas ----------
@app.get("/api/autores")
def listar_autores(db: Session = Depends(get_db)):
    # Dos consultas en total (autores + libros de todos ellos), sin importar cuántos autores haya.
    consulta = select(Autor).options(selectinload(Autor.libros)).order_by(Autor.id)
    autores = db.scalars(consulta).all()
    return {"autores": [autor_a_dict(a) for a in autores]}


@app.get("/api/autores/{autor_id}")
def obtener_autor(autor_id: int, db: Session = Depends(get_db)):
    consulta = select(Autor).options(selectinload(Autor.libros)).where(Autor.id == autor_id)
    autor = db.scalars(consulta).first()
    if autor is None:
        raise HTTPException(status_code=404, detail="Autor no encontrado")
    return autor_a_dict(autor)


@app.get("/api/libros")
def listar_libros(db: Session = Depends(get_db)):
    libros = db.scalars(select(Libro).order_by(Libro.id)).all()
    return {"libros": [libro_a_dict(l) for l in libros]}
````
